```python
from pathlib import Path
import zipfile
import urllib.request

PROJECT_ROOT = Path.cwd()

if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_DIR = PROJECT_ROOT / "data"
DATA_DIR.mkdir(parents=True, exist_ok=True)

DATASETS = {
    "ml-latest-small": {
        "url": "https://files.grouplens.org/datasets/movielens/ml-latest-small.zip",
        "zip_name": "ml-latest-small.zip",
        "required_files": ["ratings.csv", "movies.csv"],
    },
    "ml-latest": {
        "url": "https://files.grouplens.org/datasets/movielens/ml-latest.zip",
        "zip_name": "ml-latest.zip",
        "required_files": ["ratings.csv", "movies.csv"],
    },
    "ml-1m": {
        "url": "https://files.grouplens.org/datasets/movielens/ml-1m.zip",
        "zip_name": "ml-1m.zip",
        "required_files": ["ratings.dat", "movies.dat", "users.dat"],
    },
}
```

```python
def dataset_exists(dataset_name):
    dataset_dir = DATA_DIR / dataset_name
    required_files = DATASETS[dataset_name]["required_files"]

    return dataset_dir.exists() and all(
        (dataset_dir / file_name).exists()
        for file_name in required_files
    )


def download_movielens(dataset_name):
    if dataset_name not in DATASETS:
        raise ValueError(f"Unknown dataset: {dataset_name}")

    if dataset_exists(dataset_name):
        print(f"{dataset_name} already exists. Nothing to do.")
        return

    config = DATASETS[dataset_name]
    zip_path = DATA_DIR / config["zip_name"]

    if not zip_path.exists():
        print(f"Downloading {dataset_name}...")
        urllib.request.urlretrieve(config["url"], zip_path)
    else:
        print(f"{zip_path.name} already exists. Extracting...")

    with zipfile.ZipFile(zip_path, "r") as z:
        z.extractall(DATA_DIR)

    print(f"{dataset_name} is ready.")
```

```python
for dataset_name in DATASETS:
    download_movielens(dataset_name)
```
